```python
def dot(xi, w):
    n=len(xi)
    
    result =0.0
    for j in range(n):
        result = result + xi[j]*w[j]
    
    return result    
```

```python
w0 , w
```

```python
def train_linner_regression(X,y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    return w_full[0] , w_full[1:]
```

```python
train_linner_regression(X,y)
```
