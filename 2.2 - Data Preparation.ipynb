### 2.9 - RMSE

$ 
RMSE = \sqrt{\frac{1}{m} \sum_{i=1}^{m} (g(xi)-yi)^2}
$

```python
def rmse(y, y_prediction):
    error = y - y_prediction
    se = error**2
    mse = se.mean() 
    return np.sqrt(mse)
```

```python
rmse(y_train,y_prediction)
```

```python
categorical_variables = [
    'make', 'engine_fuel_type', 'transmission_type', 'driven_wheels',
    'market_category', 'vehicle_size', 'vehicle_style'
]
```

```python
categories = {}

for c in categorical_variables:
    categories[c] = list(df_train[c].value_counts().head().index)
```

```python
def prepare_X (data):
    
    data = data.copy()
    
    data["age"] = 2017 - data.year
    feachers = base + ["age"]
    
    
    for c, values in categories.items():
        for v in values:
            data['%s_%s' % (c, v)] = (data[c] == v).astype('int')
            feachers.append('%s_%s' % (c, v))
        
    
    for v in [2,3,4]:
        data['num_doors_%s'%v ]=(data.number_of_doors== v ).astype("int")
        feachers.append("num_doors_%s" %v) 
        
        
    for v in makes:
        data['make_%s'%v ]=(data.make == v ).astype("int")
        feachers.append("make_%s" %v) 
    
    data_num = data[feachers]
    data_num = data_num.fillna(0)
    X = data_num.values
    return X
```

```python
XTX = [
    [1, 2, 2],
    [2, 1, 1.00000001],
    [2, 1.00000001, 1],
] # add a small num to the diagnal to fix
XTX_inv = np.linalg.inv(XTX)
```

```python
# # add a small num to the diagnal to fix
# np.eye(3)
```

```python
XTX = XTX + 0.01 * np.eye(3)
# + 0.01 is the number added to the daiaginal 
XTX
```

```python
np.linalg.inv(XTX)
```

```python
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])

    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```

```python
X_train =  prepare_X (df_train)
w0, w = train_linear_regression_reg(X_train,y_train,r=0.001) 


X_val = prepare_X(df_val)
y_prediction=w0 + X_val.dot(w)

rmse(y_val,y_prediction)
```
